```
from google.colab import files
import pandas as pd
from itertools import product

# Upload CSV file
uploaded = files.upload()

# Get uploaded file name
filename = list(uploaded.keys())[0]

# Read CSV
data = pd.read_csv(filename)

print("Training Data:")
display(data)

# Separate attributes and target
attributes = list(data.columns[:-1])
target = data.columns[-1]

# Get possible values for each attribute
domains = [list(data[attr].unique()) for attr in attributes]

# Add '?' meaning any value
domains = [values + ['?'] for values in domains]

# Generate all possible hypotheses
all_hypotheses = list(product(*domains))


# Check whether hypothesis correctly classifies an example
def covers(hypothesis, example):
    for h, value in zip(hypothesis, example):
        if h != '?' and h != value:
            return False
    return True


# Find all hypotheses consistent with training data
consistent_hypotheses = []

for hypothesis in all_hypotheses:
    consistent = True

    for _, row in data.iterrows():
        example = row[attributes].tolist()
        actual_class = row[target]

        prediction = "Yes" if covers(hypothesis, example) else "No"

        if prediction != actual_class:
            consistent = False
            break

    if consistent:
        consistent_hypotheses.append(hypothesis)


# Display result
print("\nAll Consistent Hypotheses:")
for h in consistent_hypotheses:
    print(h)

# Find Specific Boundary
specific_boundary = []

for h in consistent_hypotheses:
    if not any(
        all(a == '?' or a == b for a, b in zip(h, other))
        and h != other
        for other in consistent_hypotheses
    ):
        specific_boundary.append(h)

# Find General Boundary
general_boundary = []

for h in consistent_hypotheses:
    if not any(
        all(a == '?' or a == b for a, b in zip(other, h))
        and h != other
        for other in consistent_hypotheses
    ):
        general_boundary.append(h)

print("\nSpecific Boundary (S):")
for h in specific_boundary:
    print(h)

print("\nGeneral Boundary (G):")
for h in general_boundary:
    print(h)
```
